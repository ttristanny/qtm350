1) List drivers ordered by `team` alphabetically and then by `victories` in descending order.

```python
query1 = """
SELECT driver_id, driver_name, team, victories
FROM drivers
ORDER BY team ASC, victories DESC;
"""

df_task1 = pd.read_sql_query(query1, conn)
print(df_task1)
```

2) List drivers who have won more races than the average number of victories across all drivers.

```python
query2 = """
SELECT driver_id, driver_name, team, victories
FROM drivers
WHERE victories > (SELECT AVG(victories) FROM drivers)
ORDER BY victories DESC;
"""

df_task2 = pd.read_sql_query(query2, conn)
print(df_task2)
```

3) Find all drivers with more than 50 podiums OR who drive for the 'Ferrari' team.

```python
query3 = """
SELECT driver_id, driver_name, team, podiums
FROM drivers
WHERE podiums > 50 OR team = 'Ferrari'
ORDER BY podiums DESC;
"""

df_task3 = pd.read_sql_query(query3, conn)
print(df_task3)
```

4) Count the number of drivers for each `nationality`, but only include drivers who have more than 40 `podiums`. Order the result by count in descending order.

```python
query4 = """
SELECT nationality, COUNT(*) as driver_count
FROM drivers
WHERE podiums > 40
GROUP BY nationality
ORDER BY driver_count DESC;
"""

df_task4 = pd.read_sql_query(query4, conn)
print(df_task4)
```

5) Calculate the average `points` for each `team`, rounded to two decimal places.

```python
query5 = """
SELECT team, ROUND(AVG(points), 2) as avg_points
FROM drivers
GROUP BY team
ORDER BY avg_points DESC;
"""

df_task5 = pd.read_sql_query(query5, conn)
print(df_task5)
```

6) List drivers who are in the top 20% based on their `points`.

```python
query6 = """
SELECT driver_id, driver_name, points
FROM drivers
WHERE points >= (
    SELECT AVG(points) + (MAX(points) - MIN(points)) * 0.3
    FROM drivers
)
ORDER BY points DESC;
"""

df_task6 = pd.read_sql_query(query6, conn)
print(df_task6)
```

7) Calculate the total `victories`, total `podiums`, and average `points` for drivers with 'German' `nationality`.

```python
query7 = """
SELECT SUM(victories) as total_victories, SUM(podiums) as total_podiums, ROUND(AVG(points), 2) as avg_points
FROM drivers
WHERE nationality = 'German';
"""

df_task7 = pd.read_sql_query(query7, conn)
print(df_task7)
```

8) List all drivers whose `driver_name` contains both the letter 'a' and the letter 'e' (case-insensitive).

```python
query8 = """
SELECT driver_id, driver_name
FROM drivers
WHERE LOWER(driver_name) LIKE '%a%' AND LOWER(driver_name) LIKE '%e%'
ORDER BY driver_name;
"""

df_task8 = pd.read_sql_query(query8, conn)
print(df_task8)
```

9) Retrieve the driver with the most `victories` within each `team`.

```python
query9 = """
SELECT driver_id, driver_name, team, victories
FROM drivers
WHERE (driver_id, victories) IN (
    SELECT driver_id, MAX(victories)
    FROM drivers
    GROUP BY team
)
ORDER BY team;
"""

df_task9 = pd.read_sql_query(query9, conn)
print(df_task9)
```

10) List drivers whose `points` are below the average `points` for their respective `team`.

```python
query10 = """
SELECT d.driver_id, d.driver_name, d.team, d.points
FROM drivers d
WHERE d.points < (
    SELECT AVG(points)
    FROM drivers
    WHERE team = d.team
)
ORDER BY d.team, d.points DESC;
"""

df_task10 = pd.read_sql_query(query10, conn)
print(df_task10)
```
